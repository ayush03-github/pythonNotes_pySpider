find first two maximum element in the list and display its difference 

```python
l1 = [1, 3, 6, 8, 11, 4, 15, 3, 7, 9, 5]

max1 = l1[0]
max2 = l1[1]

for i in range(len(l1)):
    if max1 < l1[i]:
        max2 = max1
        max1 = l1[i]
    elif max2 < l1[i]:
        max2 = l1[i]

print('first max value is :' , max1)
print('second max value is :' , max2)
print('difference of both max values is ',max1 - max2)
```

WAP to display the missing elements in a list excluding the existing values 

```python
l1 = [2,9,10,15,19,24,35]

for i in range(len(l1)-1):
    fe = l1[i]+1
    le = l1[i +1]
    for j in range(fe, le):
        print(j, end=" ")
```

WAP to display sum of all the prime numbers present in a given list using functions

```python
def isPrime(num):
    if num <= 1:
        return False
    else:
        for i in range(2, num):
            if num % i == 0:
                return False
            return True
        
l1 = [2, 11 ,15, 7, 1, 18, 25, 19, 31, 30, 44, 51]

res = 0

for i in l1:
    if isPrime(i):
        res += i

print(f"sum of prime numbers is {res}")
```

push to right n times

```python
l1 = [11,22,33,44,55]
print("original list", l1)
print()

n = 4

for _ in range(n):
    temp = l1[len(l1)-1]
    for i in range(len(l1)-1,0,-1):
        l1[i] = l1[i-1]
    l1[0] = temp
    print(l1)
```
